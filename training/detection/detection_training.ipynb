# 🚀 YOLOv8 Detecção de Objetos - Treinamento

**Instruções:**
1. Copie seu dataset (estrutura: images/train, labels/train, data.yaml) para a pasta `dataset/`
2. Execute as células em ordem
3. Certifique-se que o dataset é para DETECÇÃO (não segmentação)

```python
from ultralytics import YOLO
from pathlib import Path

# Setup de diretórios
BASE_DIR = Path.cwd()
DATASET_PATH = BASE_DIR / 'dataset'

print(f"✓ Dataset path: {DATASET_PATH}")
print(f"✓ Arquivos: {list(DATASET_PATH.glob('*'))[:5]}")
```

## 🎯 TREINAR DETECÇÃO DE OBJETOS

```python
# Parâmetros de treinamento
MODEL = 'yolov8n'           # nano, small, medium, large, xlarge
EPOCHS = 100
BATCH_SIZE = 4
IMG_SIZE = 640
DEVICE = 0                  # 0=GPU, -1=CPU
SAVE_NAME = 'detect_model'
PATIENCE = 20
SINGLE_CLASS = True        # True = unificar todas as classes em uma só

print(f"🚀 Treinando {MODEL}...")
print(f"📂 Dataset: {DATASET_PATH}")
print(f"⏰ Épocas: {EPOCHS}")
print(f"📦 Batch: {BATCH_SIZE}\n")

model = YOLO(MODEL)
results = model.train(
    data=str(DATASET_PATH / 'data.yaml'),
    epochs=EPOCHS,
    batch=BATCH_SIZE,
    imgsz=IMG_SIZE,
    device=DEVICE,
    patience=PATIENCE,
    project='runs',
    name=SAVE_NAME,
    exist_ok=False,
    single_cls=SINGLE_CLASS
)

print(f"\n✅ Treinamento concluído!")
print(f"📁 Modelo: runs/{SAVE_NAME}/weights/best.pt")
```
