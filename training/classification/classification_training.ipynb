# 🚀 YOLOv8 Classificação - Treinamento

**Instruções:**
1. Copie seu dataset (estrutura: train/class1, train/class2, val/class1, val/class2) para a pasta `dataset/`
2. Execute as células em ordem

```python
from ultralytics import YOLO
from pathlib import Path

# Setup de diretórios
BASE_DIR = Path.cwd()
DATASET_PATH = BASE_DIR / 'dataset'

print(f"✓ Dataset path: {DATASET_PATH}")
print(f"✓ Arquivos: {list(DATASET_PATH.glob('*'))[:5]}")
```

## 🎯 TREINAR CLASSIFICAÇÃO

```python
# Parâmetros de treinamento
MODEL = 'yolov8n-cls'        # nano, small, medium, large, xlarge
EPOCHS = 100
BATCH_SIZE = 8
IMG_SIZE = 640
DEVICE = 0                  # 0=GPU, -1=CPU
SAVE_NAME = 'dedos'

print(f"🚀 Treinando {MODEL}...")
print(f"📂 Dataset: {DATASET_PATH}")
print(f"⏰ Épocas: {EPOCHS}")
print(f"📦 Batch: {BATCH_SIZE}\n")

model = YOLO(MODEL)
results = model.train(
    data=str(DATASET_PATH),
    epochs=EPOCHS,
    batch=BATCH_SIZE,
    imgsz=IMG_SIZE,
    device=DEVICE,
    patience=10,
    project='runs',
    name=SAVE_NAME,
    exist_ok=False
)

print(f"\n✅ Treinamento concluído!")
print(f"📁 Modelo: runs/{SAVE_NAME}/weights/best.pt")
```
